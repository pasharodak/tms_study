```python
import os
import time
import shutil
from spleeter.separator import Separator

INPUT_FILE = "data/3.mp3"
OUT_DIR = "data/3_spleeter"

os.makedirs(OUT_DIR, exist_ok=True)

def separate_with_spleeter():
    print("⚡ Spleeter (2 stems)")

    start = time.time()

    separator = Separator("spleeter:2stems")
    separator.separate_to_file(INPUT_FILE, OUT_DIR)

    elapsed = time.time() - start
    print(f"✅ Готово за {elapsed:.2f} сек")

    base = os.path.splitext(os.path.basename(INPUT_FILE))[0]
    src_dir = os.path.join(OUT_DIR, base)

    shutil.copy2(
        os.path.join(src_dir, "vocals.wav"),
        os.path.join(OUT_DIR, "vocals.wav")
    )
    shutil.copy2(
        os.path.join(src_dir, "accompaniment.wav"),
        os.path.join(OUT_DIR, "music.wav")
    )

    print("🎵 vocals.wav")
    print("🎵 music.wav")

if __name__ == "__main__":
    separate_with_spleeter()
```
